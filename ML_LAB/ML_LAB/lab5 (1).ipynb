4.Create a DataFrame containing the columns Name, Gender, and Age, and perform
DataFrame slicing to:
o Display the first two rows.
o Display all rows from the third row onwards.

```python
data2 = {
    "Name": ["Devu", "Gladia", "Nikhitha", "Gowri","Merryl"],
    "Gender": ["Female", "Female", "Female", "Female", "Female"],
    "Age": [21, 22, 23, 20, 24]
}
people = pd.DataFrame(data2)
print("\nOriginal DataFrame:")
print(people)
print("\nFirst Two Rows:")
print(people.iloc[:2])
print("\nThird Row Onwards:")
print(people.iloc[2:])
```

5.Create a Pandas Series containing four random numbers, and display:
o The complete Series.
o The values stored in the Series.
o The first two elements using head().
o The last three elements using tail().

```python
random_series = pd.Series(np.random.rand(4))
print("\nComplete Series:")
print(random_series)
print("\nValues:")
print(random_series.values)
print("\nFirst Two Elements:")
print(random_series.head(2))
print("\nLast Three Elements:")
print(random_series.tail(3))
```

6. Create a DataFrame containing the columns Name, Age, and Rating, and display the
following DataFrame properties:
o Transpose of the DataFrame.
o Row and column labels using axes.
o Data type of each column using dtypes.
o Whether the DataFrame is empty using empty.
o Number of dimensions using ndim.
o Shape of the DataFrame using shape.
o Total number of elements using size.
o Values stored in the DataFrame using values.
o Number of missing values in each column using isnull().sum().

```python
data3 = {
    "Name": ["Devu", "Gladia", "Nikhitha", "Gowri"],
    "Age": [21, 22, 23, 20],
    "Rating": [4.5, 4.0, 4.8, 4.2]
}
df = pd.DataFrame(data3)
print("\nOriginal DataFrame:")
print(df)
print("\nTranspose:")
print(df.T)
print("\nAxes:")
print(df.axes)
print("\nData Types:")
print(df.dtypes)
print("\nIs DataFrame Empty?")
print(df.empty)
print("\nNumber of Dimensions:")
print(df.ndim)
print("\nShape:")
print(df.shape)
print("\nSize:")
print(df.size)
print("\nValues:")
print(df.values)
print("\nMissing Values:")
print(df.isnull().sum())
```
