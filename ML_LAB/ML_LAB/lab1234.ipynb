1.Draw a line in a diagram from position (1, 3) to (2, 10) then to (6, 12) and finally to position (18, 20).
(Mark each point with a beautiful green colour and set line colour to red and line style dotted).

```python
import matplotlib.pyplot as plt
x = [1, 2, 6, 18]
y = [3, 10, 12, 20]
plt.figure(figsize=(8, 6))
plt.plot(x, y, color='red', linestyle='dotted', marker='o', markerfacecolor='green', markeredgecolor='green')
plt.title('Diagram with Dotted Red Line and Green Markers')
plt.xlabel('X position')
plt.ylabel('Y position')
plt.show()
```

c) Create a scatter plot of the Iris dataset using different colors for each flower variety (Setosa,
Versicolor, and Virginica).

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")
plt.figure(figsize=(8, 6))
plt.scatter(
    df[df["species"] == "setosa"]["sepal_length"],
    df[df["species"] == "setosa"]["petal_length"],
    color="red",
    label="Setosa"
)
plt.scatter(
    df[df["species"] == "versicolor"]["sepal_length"],
    df[df["species"] == "versicolor"]["petal_length"],
    color="green",
    label="Versicolor"
)
plt.scatter(
    df[df["species"] == "virginica"]["sepal_length"],
    df[df["species"] == "virginica"]["petal_length"],
    color="blue",
    label="Virginica"
)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Iris Dataset Scatter Plot")
plt.legend()
plt.grid(True)
plt.show()
```

d) Create a line plot showing the relationship between sepal.length and sepal.width.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")
plt.figure(figsize=(8, 6))
plt.plot(
    df["sepal_length"],
    df["sepal_width"],
    marker="o",
    color="blue"
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Relationship Between Sepal Length and Sepal Width")
plt.grid(True)
plt.show()
```

e) Create a histogram showing the distribution of sepal.length values.

```python
import matplotlib.pyplot as plt
plt.hist(df["sepal_length"], bins=10, edgecolor="black")
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

f) Create a bar chart showing the total sepal.length for each Iris flower variety.

```python
import matplotlib.pyplot as plt
total_sepal = df.groupby("species")["sepal_length"].sum()
total_sepal.plot(kind="bar", edgecolor="black")
plt.title("Total Sepal Length by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Total Sepal Length")
plt.xticks(rotation=0)
plt.show()
```

g) Create a box plot to visualize the distribution of sepal.length for different Iris flower varieties.

```python
import matplotlib.pyplot as plt
df.boxplot(column="sepal_length", by="species")
plt.title("Distribution of Sepal Length by Iris Species")
plt.suptitle("")
plt.xlabel("Iris Species")
plt.ylabel("Sepal Length")
plt.show()
```

h) Create a pie chart showing the percentage distribution of the three Iris flower varieties in the
dataset.

```python
import matplotlib.pyplot as plt

species_count = df["species"].value_counts()

plt.pie(species_count,
        labels=species_count.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Percentage Distribution of Iris Flower Varieties")
plt.show()
```

i) Create a scatter plot comparing petal.length and petal.width, using different colors for each Iris
variety.

```python
import matplotlib.pyplot as plt
for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(subset["petal_length"],
                subset["petal_width"],
                label=species)
plt.title("Petal Length vs Petal Width by Iris Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.legend()
plt.show()
```

j) Create multiple subplots displaying the following histograms:

Histogram of sepal.length
Histogram of sepal.width
Histogram of petal.length
Histogram of petal.width

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Histogram of sepal length
axes[0, 0].hist(df["sepal_length"], bins=10, edgecolor="black")
axes[0, 0].set_title("Histogram of Sepal Length")
axes[0, 0].set_xlabel("Sepal Length")
axes[0, 0].set_ylabel("Frequency")

# Histogram of sepal width
axes[0, 1].hist(df["sepal_width"], bins=10, edgecolor="black")
axes[0, 1].set_title("Histogram of Sepal Width")
axes[0, 1].set_xlabel("Sepal Width")
axes[0, 1].set_ylabel("Frequency")

# Histogram of petal length
axes[1, 0].hist(df["petal_length"], bins=10, edgecolor="black")
axes[1, 0].set_title("Histogram of Petal Length")
axes[1, 0].set_xlabel("Petal Length")
axes[1, 0].set_ylabel("Frequency")

# Histogram of petal width
axes[1, 1].hist(df["petal_width"], bins=10, edgecolor="black")
axes[1, 1].set_title("Histogram of Petal Width")
axes[1, 1].set_xlabel("Petal Width")
axes[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

k) Create a grouped bar chart comparing the average sepal.length and sepal.width for each Iris
flower variety.

```python
import matplotlib.pyplot as plt
avg_values = df.groupby("species")[["sepal_length", "sepal_width"]].mean()
avg_values.plot(kind="bar", edgecolor="black")
plt.title("Average Sepal Length and Sepal Width by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend(["Sepal Length", "Sepal Width"])
plt.show()
```

l) Create a bubble scatter plot using petal.length and petal.width, where the size of each bubble
represents sepal.length.

```python
import matplotlib.pyplot as plt
plt.scatter(
    df["petal_length"],
    df["petal_width"],
    s=df["sepal_length"] * 50,
    alpha=0.6
)
plt.title("Bubble Scatter Plot")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```
